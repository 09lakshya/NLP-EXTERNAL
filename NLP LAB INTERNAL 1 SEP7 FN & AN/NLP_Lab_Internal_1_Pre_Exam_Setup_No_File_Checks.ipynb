## 5. Check NLTK resources

```python
checks = [
    ('punkt', 'tokenizers/punkt'),
    ('punkt_tab', 'tokenizers/punkt_tab'),
    ('stopwords', 'corpora/stopwords'),
    ('wordnet', 'corpora/wordnet'),
    ('omw-1.4', 'corpora/omw-1.4'),
    ('averaged_perceptron_tagger', 'taggers/averaged_perceptron_tagger')
]

for name, path in checks:
    try:
        location = nltk.data.find(path)
        print(f'OK   {name}: {location}')
    except LookupError:
        print(f'FAIL {name}: resource not found')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords, wordnet

print('\nTokenization test:', word_tokenize('NLP lab setup is ready.'))
print('Stopword test:', stopwords.words('english')[:10])
print('WordNet test:', wordnet.synsets('computer')[:2])
```

## 6. Check spaCy model

```python
import spacy

nlp = spacy.load('en_core_web_sm')
doc = nlp('The students are learning Natural Language Processing.')

for token in doc:
    print(token.text, '-> POS:', token.pos_, '| Lemma:', token.lemma_, '| Dep:', token.dep_)

print('\nspaCy dependency/POS/lemma test completed.')
```
